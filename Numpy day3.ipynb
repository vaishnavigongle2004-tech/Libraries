5 employee ki salary mein 10% increment karna hai

```python
salary=[20000,25000,30000,35000,40000]
new_salary=[]
for s in salary:
    new_salary.append(s*1.10)
print(new_salary)
```

```python
#numpy vectorisation
salary=np.array([20000,25000,30000,35000,40000])
new_salary=salary*1.10
print(new_salary)
```

```python
#salary increment
salary=np.array([20000,25000,30000,35000,40000])
Increment_salary=salary+5000
print(Increment_salary)
```

```python
#product discount
prices=np.array([1000,2000,3000,4000])
discounted_prices=prices*0.90 #10% discount apply
print(discounted_prices)
```

```python
#students marks calculation
marks=np.array([70,89,65,89,85])
add_marks=marks+5
print(add_marks)
```

```python
#temperature conversion
celsius=np.array([0,10,20,30,40])
fahrenheit=(celsius*9/5)+32
print(fahrenheit)
```
