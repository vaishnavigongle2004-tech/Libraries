broadcasting with two array 

```python
prices=np.array([100,200,300])
discount=np.array([10,20,30])
result=prices-discount
result
```

numpy broadcasting ke liye dimentions ko right side se compare karega
dono dimension equal chaiye
ye kisi dimension ki value 1 ho
array A      array B    broadcasting
(3,3)         (3,3)      possible
(3,3)         (1,3)      possible
(3,3)         (3,1)      possible
(3,3)         (2,3)      possible
(3,3)         (2,3)      not possible
