```python
import numpy as np
```

Indexing

```python
fruits=np.array(["Apple","Banana","Grapes","Orange","Strawberry"])
print(fruits[0])
print(fruits[1])
print(fruits[2])
```

Negative Indexing

```python
fruits=np.array(["Apple","Banana","Grapes","Orange","Strawberry"])
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
print(fruits[-4])
print(fruits[-5])
```

```python
Slicing
```

```python
print(fruits[1:4])
print(fruits[2:4])
print(fruits[-1:-4])
```

2d Arrray

```python
marks=np.array([
    [80,85],
    [90,85],
])
print(marks[0])
print(marks[1])
```

Addition

```python
a=np.array([10,20,30])
b=np.array([1,2,3])
print(a+b)
```

Comparision-operator

```python
marks=np.array([35,55,70,40,90])
print(marks>50)
print(marks<50)
print(marks==70)
```

Boolean masking


```python
marks=np.array([35,55,70,40,90])
mask=marks>50
print(mask)
print(marks[mask])
```

Student marks report

```python
import numpy as np

names=np.array(["Aman","Riya","Sahil","Priya","Neha"])
marks=np.array([35,55,70,40,90])

passed=marks>=40

print("Passed students:",names[passed])
print("Pass marks:",marks[passed])
print("Failed students:",names[~passed])
```
